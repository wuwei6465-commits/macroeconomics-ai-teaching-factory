cell1:财政政策与挤出效应

### 扩张性财政政策与挤出效应

**1. 政策冲击（IS 曲线右移）**
* 政府支出 $\Delta G$ 增加，乘数效应驱动总需求扩张。
* 理想状态（若利率不变）：产出将达到“理论最大增量”。

**2. 利率攀升（沿 LM 曲线移动）**
* 国民收入 $Y$ 增加 $\to$ 货币的交易性需求上升。
* 货币供给固定 $\to$ 导致市场利率 $r$ 被迫推高。

**3. 挤出效应（Crowding-Out Effect）**
* 利率 $r$ 上升，抬高了融资成本，从而抑制私人投资。
* **最终结果**：实际产出增幅 $<$ 理论最大增幅（差额即为被“挤出”的部分）。

---
💡 **课堂实验：**
* 拖动 **政府购买增量**：观察新均衡点（红蓝交点）的移动轨迹。
* 调节 **货币需求敏感度**：思考斜率变化将如何放大或缩小挤出效应？

```python
import numpy as np
import plotly.graph_objects as go
from ipywidgets import interact, FloatSlider

@interact(dG=FloatSlider(min=0, max=200, step=10, value=100, description='政府购买增量'),
          h=FloatSlider(min=10, max=100, step=10, value=50, description='货币需求敏感度'))
def cell_1_fiscal_policy(dG, h):
    Y = np.linspace(400, 1600, 100)
    # 基础参数
    MPC, T, I_base, d, k, M_P = 0.8, 200, 300, 20, 0.5, 400
    multiplier = 1 / (1 - MPC)

    # IS1 (初始) 与 IS2 (冲击后)
    r_is1 = (200 - MPC*T + I_base - (1-MPC)*Y) / d
    r_is2 = (200 + dG - MPC*T + I_base - (1-MPC)*Y) / d
    # LM
    r_lm = (k*Y - M_P) / h

    fig = go.Figure()
    fig.add_trace(go.Scatter(x=Y, y=r_is1, name='初始 IS', line=dict(color='gray', dash='dot')))
    fig.add_trace(go.Scatter(x=Y, y=r_is2, name='扩张后 IS', line=dict(color='red', width=3)))
    fig.add_trace(go.Scatter(x=Y, y=r_lm, name='LM 曲线', line=dict(color='blue', width=3)))

    # 标注理论最大位移 (凯恩斯交叉乘数)
    theoretical_Y = 1000 + dG * multiplier
    
    # 优化点 1：将标注的 y 坐标稍微抬高至 0.5，避免箭头和文字被 X 轴遮挡
    fig.add_annotation(x=theoretical_Y, y=0.5, text="理论最大增量", showarrow=True, arrowhead=1)

    fig.update_layout(
        title=f"单元 1: 财政政策与挤出效应 (位移: {dG*multiplier:.1f})",
        xaxis_title="产出 Y",
        yaxis_title="利率 r (%)",
        template="plotly_white",
        
        # 优化点 2：固定图表的物理宽高，防止随浏览器窗口随意拉伸
        width=750,   
        height=550,  
        
        # 优化点 3：微调四周边缘留白（Margin），让图表主体更加饱满
        margin=dict(l=60, r=40, t=60, b=50), 
        
        xaxis=dict(range=[600, 1600], showline=True, linecolor='black'),
        yaxis=dict(range=[0, 15], showline=True, linecolor='black')
    )
    fig.show()
```

Cell 2：货币传导机制——流动性的渗透，针对难点：观察M↑→r↓→I↑→Y↑ 的完整链条。

### 货币传导机制与扩张性货币政策

**1. 政策冲击（LM 曲线右移）**
* 央行增加货币供给量 $\Delta M$。
* 资金面变得宽松，货币市场出现超额供给，导致 **市场利率 $r$ 下降**。
* **视觉呈现**：“初始 LM”（灰色虚线）向右下方平移至“扩张后 LM”（蓝色实线）。

**2. 投资刺激与产出增加（沿 IS 曲线移动）**
* 利率 $r$ 下降 $\to$ 企业融资成本降低 $\to$ **刺激私人投资 $I$** 增加。
* 投资增加拉动总需求，通过乘数效应最终带动 **国民收入 $Y$ 上升**。
* **视觉呈现**：宏观经济均衡点沿着“IS 曲线”（红色实线）向右下方滑动。

**3. 最终结论**
* 扩张性货币政策的标准传导结果：**利率降低，产出增加**。

---
💡 **课堂实验：**
* 拖动上方 **“货币供给增量”** 滑块：观察 LM 曲线的位移以及红蓝交点（新均衡点）的运动轨迹。
* **深度思考**：如果红色的 IS 曲线变得极其陡峭（即企业的投资对利率变化**非常不敏感**），货币政策拉动经济的效果是会变好，还是变差？为什么？

```python
import numpy as np
import plotly.graph_objects as go
from ipywidgets import interact, FloatSlider

@interact(dM=FloatSlider(min=0, max=200, step=10, value=100, description='货币供给增量'))
def cell_2_monetary_policy(dM):
    # 将 X 轴计算范围稍微拉宽，与单元 1 保持一致，确保曲线画满整个坐标系
    Y = np.linspace(600, 1600, 100) 
    MPC, T, G, I_base, d, k, M_P_base, h = 0.8, 200, 200, 300, 20, 0.5, 400, 50

    r_is = (G - MPC*T + I_base - (1-MPC)*Y) / d
    r_lm1 = (k*Y - M_P_base) / h
    r_lm2 = (k*Y - (M_P_base + dM)) / h

    fig = go.Figure()
    fig.add_trace(go.Scatter(x=Y, y=r_is, name='IS 曲线', line=dict(color='red', width=3)))
    fig.add_trace(go.Scatter(x=Y, y=r_lm1, name='初始 LM', line=dict(color='gray', dash='dot')))
    fig.add_trace(go.Scatter(x=Y, y=r_lm2, name='扩张后 LM', line=dict(color='blue', width=3)))

    fig.update_layout(
        title="单元 2: 货币传导机制 (货币扩张导致利率下降并刺激总产出)",
        xaxis_title="产出 Y", 
        yaxis_title="利率 r (%)", 
        template="plotly_white",
        
        # 强制设定物理尺寸，确保长宽比协调
        width=750,
        height=550,
        # 优化四周留白
        margin=dict(l=60, r=40, t=60, b=50),
        
        # 锁定坐标轴范围。
        # 这一步非常重要，否则拖动滑块时整个坐标系会跟着缩放跳动，影响观察
        xaxis=dict(range=[600, 1600], showline=True, linecolor='black'),
        yaxis=dict(range=[0, 15], showline=True, linecolor='black')
    )
    fig.show()
```

Cell 3：政策互动模拟器 (央行的三副面孔)针对难点：应对财政扩张时，美联储选择“目标M”、“目标r”或“目标Y”的不同后果。

### 政策博弈：央行如何应对财政扩张？

当政府踩下油门（IS 右移）时，宏观经济的最终走向取决于央行的反应：

| 美联储目标 | 货币政策操作 (央行响应) | 图形轨迹 (LM 曲线) | 最终经济后果 (挤出效应) |
| :--- | :--- | :--- | :--- |
| **保持 M 不变** | 保持中立（供给不变） | 🔵 **原位不动** | $Y$ 增加，$r$ 上升（**部分挤出**） |
| **保持 r 不变** | 货币适应（同步放水） | 🔵 **向右下平移** | $Y$ 增幅最大，$r$ 不变（**零挤出**） |
| **保持 Y 不变** | 主动紧缩（踩下刹车） | 🔵 **向左上平移** | $Y$ 锁死原地，$r$ 飙升（**完全挤出**） |

---
💡 **沙盘推演：**
先拉大 **财政扩张 dG**，然后切换下拉菜单，观察蓝线 (LM) 是如何“配合”或“阻截”红线 (IS) 的。

```python
import numpy as np
import plotly.graph_objects as go
from ipywidgets import interact, FloatSlider, Dropdown

@interact(Fed_Target=Dropdown(options=['保持 M 不变', '保持 r 不变', '保持 Y 不变'], description='美联储目标'),
          dG=FloatSlider(min=0, max=100, step=10, value=50, description='财政扩张 dG'))
def cell_3_policy_interaction(Fed_Target, dG):
    # 扩大 X 轴计算范围，以容纳完全配合下的最大产出
    Y = np.linspace(1000, 1850, 200) 
    
    # 基础参数
    MPC, T, G_base, I_base, d, k, M_base, h = 0.8, 200, 200, 300, 20, 0.5, 400, 50
    
    # ---------------- 经济学逻辑：动态计算均衡点 ----------------
    IS_intercept = (G_base - MPC*T + I_base) / d
    IS_slope = (1 - MPC) / d
    LM_intercept = - M_base / h
    LM_slope = k / h
    
    # 计算初始均衡 (E0)
    Y_init = (IS_intercept - LM_intercept) / (IS_slope + LM_slope)
    r_init = IS_intercept - IS_slope * Y_init

    # 初始 IS 与 扩张后新 IS 曲线
    r_is_init = IS_intercept - IS_slope * Y
    r_is_new = (G_base + dG - MPC*T + I_base) / d - IS_slope * Y

    fig = go.Figure()

    # 1. 绘制初始状态参考线
    fig.add_trace(go.Scatter(x=Y, y=r_is_init, name='初始 IS', line=dict(color='gray', dash='dot', width=2)))
    r_lm_init = LM_slope * Y + LM_intercept
    fig.add_trace(go.Scatter(x=Y, y=r_lm_init, name='初始 LM', line=dict(color='lightgray', dash='dot', width=2)))

    # 2. 根据美联储目标确定 LM 响应行为与新均衡点 (E1)
    if Fed_Target == '保持 M 不变':
        r_lm_new = r_lm_init
        title_suffix = " (挤出效应存在)"
        Y_new = ((G_base + dG - MPC*T + I_base) / d - LM_intercept) / (IS_slope + LM_slope)
        r_new = LM_slope * Y_new + LM_intercept
        
    elif Fed_Target == '保持 r 不变':
        r_lm_new = np.full_like(Y, r_init)
        title_suffix = " (完全配合，无挤出)"
        Y_new = ((G_base + dG - MPC*T + I_base) / d - r_init) / IS_slope
        r_new = r_init
        
    else: # 保持 Y 不变
        # 模拟陡峭的 LM 曲线强制锁定在初始 Y
        r_lm_new = 100 * (Y - Y_init) + r_init 
        title_suffix = " (完全对抗，100% 挤出)"
        Y_new = Y_init
        r_new = (G_base + dG - MPC*T + I_base) / d - IS_slope * Y_new

    fig.add_trace(go.Scatter(x=Y, y=r_is_new, name='扩张后 IS', line=dict(color='red', width=3)))
    fig.add_trace(go.Scatter(x=Y, y=r_lm_new, name='LM 响应', line=dict(color='blue', width=3)))

    # 3. 绘制均衡点标记 (E0 和 E1)，增强教学直观性
    fig.add_trace(go.Scatter(x=[Y_init], y=[r_init], mode='markers', name='初始均衡 E0', 
                             marker=dict(color='black', size=10)))
    fig.add_trace(go.Scatter(x=[Y_new], y=[r_new], mode='markers', name='新均衡 E1', 
                             marker=dict(color='red', size=12, symbol='star')))

    # 4. 严格执行视觉与学术排版规范
    fig.update_layout(
        title=f"单元 3: 政策博弈系统{title_suffix}",
        xaxis_title="国民收入 Y",
        yaxis_title="利率 r (%)",
        template="plotly_white",
        
        # 锁定固定尺寸
        width=750,
        height=550,
        margin=dict(l=60, r=40, t=60, b=50),
        
        # 将图例移至左上角内部，避免挤压主图空间
        legend=dict(yanchor="top", y=0.98, xanchor="left", x=0.02, bgcolor="rgba(255, 255, 255, 0.8)"),
        
        # 锁定坐标轴范围，X轴上限拉长至 1850，确保完全配合模式下图线不断裂
        xaxis=dict(range=[1000, 1850], showline=True, linecolor='black', linewidth=1, showgrid=False, zeroline=False),
        yaxis=dict(range=[0, 15], showline=True, linecolor='black', linewidth=1, showgrid=False, zeroline=False)
    )
    fig.show()
```

Cell 4：终极推导——从 IS-LM 绘制 AD 曲线。针对难点：通过变动物价P观察LM位移，并在下方坐标系中自动生成AD点。

### 宏观经济的连通器：推导总需求 (AD) 曲线

总需求 (AD) 曲线并不是凭空存在的，它本质上是 **IS-LM 模型在不同物价水平下的均衡点集合**。

理解 AD 曲线为何向右下方倾斜，核心在于掌握物价变动引发的“**实际货币余额效应**”。请对照图表观察以下传导链条：

| 物价水平 ($P$) | 实际货币供给 ($M/P$) | LM 曲线轨迹 (上图) | 利率 ($r$) | 最终产出 ($Y$) | AD 空间映射 (下图) |
| :--- | :--- | :--- | :--- | :--- | :--- |
| 📈 **上升** | 减少 (流动性收紧) | 🔵 向**左上**移动 | 上升 | **下降** | 对应 AD 上的**左上端** |
| 📉 **下降** | 增加 (流动性宽松) | 🔵 向**右下**移动 | 下降 | **增加** | 对应 AD 上的**右下端** |

---
💡 **白盒实验：寻找消失的第三根轴**
请拖动最上方的 **“物价水平 P”** 滑块，观察上下两个坐标系的奇妙联动：
1. **驱动引擎（看上方）**：滑块的移动是如何直接“推拉”蓝色 LM 曲线的？
2. **时空投影（看下方）**：注意那条贯穿上下的**灰色垂直虚线**。上方坐标系每产生一个新均衡点，是如何将其精准投影到下方坐标系，最终“描绘”出整条绿色 AD 曲线的？

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from ipywidgets import interact, FloatSlider

@interact(P=FloatSlider(min=0.5, max=2.0, step=0.1, value=1.0, description='物价水平 P'))
def cell_4_ad_derivation_v3(P):
    # 统一 X 轴范围，这是上下图完美对齐的关键基准
    Y_range = np.linspace(600, 1600, 100)
    
    # 提取所有基础参数，避免公式中出现硬编码数字 (提取了 I_base=300)
    G, M, MPC, T, d, h, k, I_base = 200, 400, 0.8, 200, 20, 50, 0.5, 300
    
    # 动态计算当前均衡产出 Y_eq 和利率 r_eq
    Y_eq = (h*(G - MPC*T + I_base) + d*(M/P)) / (d*k + h*(1 - MPC))
    r_eq = (k*Y_eq - (M/P)) / h
    
    # 创建双面板，微调间距让视觉更紧凑
    fig = make_subplots(rows=2, cols=1, 
                        subplot_titles=("1. IS-LM 空间 (物价 P 变动改变实际货币余额)", "2. AD 空间 (推导出物价与产出的负相关)"),
                        vertical_spacing=0.12)
    
    # --- 上图: IS-LM 曲线 ---
    r_is = (G - MPC*T + I_base - (1-MPC)*Y_range) / d
    r_lm = (k*Y_range - (M/P)) / h
    
    fig.add_trace(go.Scatter(x=Y_range, y=r_is, name='IS 曲线', line=dict(color='red', width=3)), row=1, col=1)
    fig.add_trace(go.Scatter(x=Y_range, y=r_lm, name=f'LM (P={P:.1f})', line=dict(color='blue', width=3)), row=1, col=1)
    # 均衡点
    fig.add_trace(go.Scatter(x=[Y_eq], y=[r_eq], mode='markers', marker=dict(size=10, color='black'), name='当前均衡'), row=1, col=1)
    
    # --- 下图: AD 曲线 ---
    P_vals = np.linspace(0.4, 2.2, 50)
    Y_vals = (h*(G - MPC*T + I_base) + d*(M/P_vals)) / (d*k + h*(1 - MPC))
    
    fig.add_trace(go.Scatter(x=Y_vals, y=P_vals, name='AD 曲线', line=dict(color='green', width=4)), row=2, col=1)
    # 映射点 (采用菱形增强视觉区分度)
    fig.add_trace(go.Scatter(x=[Y_eq], y=[P], mode='markers', marker=dict(size=12, color='black', symbol='diamond'), name='AD 映射点'), row=2, col=1)
    
    # --- 绘制垂直映射虚线 ---
    # 上图：从均衡点垂直向下画到 X 轴
    fig.add_shape(type="line", x0=Y_eq, y0=0, x1=Y_eq, y1=r_eq, xref="x1", yref="y1", 
                  line=dict(color="gray", dash="dash", width=1.5))
    # 下图：从顶部垂直向下画到映射点
    fig.add_shape(type="line", x0=Y_eq, y0=2.2, x1=Y_eq, y1=P, xref="x2", yref="y2", 
                  line=dict(color="gray", dash="dash", width=1.5))
    
    # --- 严格的视觉与坐标轴控制 ---
    fig.update_layout(
        title_text="单元 4: 总需求 (AD) 曲线的逻辑推导",
        height=750, 
        width=750, # 强制锁定物理宽度，防止错位
        template="plotly_white", 
        showlegend=False,
        margin=dict(l=60, r=40, t=80, b=50)
    )
    
    # 【最关键的一步】强制锁定上下两个图表的 X 轴和 Y 轴范围。
    # 确保拖动滑块时，坐标系绝对静止，仅有曲线平移，呈现教科书般的严谨。
    for i in [1, 2]:
        fig.update_xaxes(title_text="产出 Y", range=[600, 1600], showline=True, linecolor='black', 
                         linewidth=1, showgrid=False, zeroline=False, row=i, col=1)
        
    fig.update_yaxes(title_text="利率 r (%)", range=[0, 15], showline=True, linecolor='black', 
                     linewidth=1, showgrid=False, zeroline=False, row=1, col=1)
    fig.update_yaxes(title_text="物价 P", range=[0.4, 2.2], showline=True, linecolor='black', 
                     linewidth=1, showgrid=False, zeroline=False, row=2, col=1)
    
    fig.show()
```

Cell 5：大萧条与流动性陷阱的极端模拟，针对难点：当利率降至0后，LM 曲线的水平形态。

### 流动性陷阱：货币政策的“失灵”时刻

当经济步入深度萧条，名义利率降至接近于零（零下限，ZLB）时，公众会预期利率已无下降空间（债券价格已达顶峰）。此时，公众对货币的投机需求趋于无限大，愿意吸收央行释放的任何新增流动性。

在图形上，LM 曲线在极低利率区间变平，形成了水平的**“流动性陷阱区”**。请对比两种状态：

| 经济所处区间 | 央行扩张操作 | LM 曲线的动态反应 | 最终宏观经济后果 |
| :--- | :--- | :--- | :--- |
| **常态区间** | 增加货币供给 $M$ | 整体向右下方平移 | $r$ 顺利下降，**$Y$ 增加** |
| **陷阱区间** | 增加货币供给 $M$ | **仅水平段向右无限延长** | $r$ 触底不变，**$Y$ 彻底锁死** |

---
💡 **极限压力测试（白盒实验）：**
1. **徒劳的央行**：请大幅向右拖动上方的 **“货币供给 M”** 滑块。观察蓝色 LM 曲线的右倾部分虽然在向右移动，但处在水平萧条段的**当前均衡点（★）**为何纹丝不动？
2. **破局思考**：既然央行“放水”已经无法拉低利率、无法刺激投资，如果你是宏观经济决策者，此时应该动用哪种政策工具才能把代表经济产出的黑星（★）向右拉出泥潭？

```python
import numpy as np
import plotly.graph_objects as go
from ipywidgets import interact, FloatSlider

@interact(M_Supply=FloatSlider(min=100, max=600, step=50, value=300, description='货币供给 M'))
def cell_5_liquidity_trap_v3(M_Supply):
    Y = np.linspace(0, 1000, 200)
    
    # 1. 模拟 LM 曲线 (带零利率下限): r = max(0, (kY - M)/h)
    k, h = 0.5, 40
    r_lm = np.maximum(0, (k * Y - M_Supply) / h)
    
    # 2. 模拟萧条期的 IS 曲线 (位置靠左且陡峭)
    # 方程: r = (100 - 0.2*Y) / 10 = 10 - 0.02Y
    r_is = 10 - 0.02 * Y
    
    # 3. 动态计算均衡点 (IS-LM 交点)
    # 当 M 足够大时，交点在 r=0 处。IS 曲线 r=0 时，Y = 10 / 0.02 = 500
    # 计算使得 r=0 的临界货币供给：M_critical = 250
    if M_Supply >= 250:
        Y_eq = 500
        r_eq = 0
    else:
        # 联立方程求解正常区域的交点: 10 - 0.02Y = 0.0125Y - M_Supply/40
        Y_eq = (10 + M_Supply / 40) / 0.0325
        r_eq = 10 - 0.02 * Y_eq

    fig = go.Figure()
    
    # 绘制 LM 曲线
    fig.add_trace(go.Scatter(x=Y, y=r_lm, name='LM (包含零下限)', line=dict(color='blue', width=4)))
    
    # 绘制 IS 曲线
    fig.add_trace(go.Scatter(x=Y, y=r_is, name='IS (萧条期)', line=dict(color='red', width=4)))
    
    # 绘制当前均衡点 (星号标记)
    fig.add_trace(go.Scatter(x=[Y_eq], y=[r_eq], mode='markers', 
                             marker=dict(size=14, color='black', symbol='star'), name='当前均衡点'))

    # 4. 优化标注位置：动态居中于平坦区域
    # 平坦段长度为 M/k，中点在 M/(2*k)，代码化简为 M_Supply
    flat_region_midpoint = M_Supply
    fig.add_annotation(x=flat_region_midpoint, y=0.5, text="流动性陷阱区", 
                       showarrow=False, font=dict(color="blue", size=14))
    
    # 5. 严格执行视觉规范
    fig.update_layout(
        title="单元 5: 流动性陷阱模拟 (当利率触及 0，增加 M 无法再增加产出)",
        xaxis_title="国民收入 Y",
        yaxis_title="利率 r (%)",  # 统一 Y 轴名称
        template="plotly_white",
        
        # 统一物理尺寸与留白
        width=750,
        height=550,
        margin=dict(l=60, r=40, t=60, b=50),
        
        # 移动图例至右上角内部
        legend=dict(yanchor="top", y=0.98, xanchor="right", x=0.98, bgcolor="rgba(255, 255, 255, 0.8)"),
        
        # 锁定坐标轴范围。Y 轴下限设为 -0.5，既能清晰展示 r=0 的平坦段，又避免底部留白过多
        xaxis=dict(showline=True, linecolor='black', linewidth=1, showgrid=False, zeroline=False, range=[0, 1000]),
        yaxis=dict(showline=True, linecolor='black', linewidth=1, showgrid=False, zeroline=False, range=[-0.5, 10])
    )
    
    fig.show()
```
